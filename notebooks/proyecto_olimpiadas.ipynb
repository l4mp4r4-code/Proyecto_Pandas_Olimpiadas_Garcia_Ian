```python
import pandas as pd
```

```python
df_atletas = pd.read_csv('../data/athlete_events.csv')
df_regiones = pd.read_csv('../data/noc_regions.csv')

print("Dimensiones atletas (filas, columnas):", df_atletas.shape)
print("Dimensiones regiones (filas, columnas):", df_regiones.shape)
```

```python
print("--- VARIABLES DE ATLETAS ---")
df_atletas.info()

print("\n--- VARIABLES DE REGIONES ---")
df_regiones.info()
```

```python
print("--- ESTADÍSTICAS BÁSICAS DE ATLETAS ---")
display(df_atletas.describe(include='all'))

print("\n--- ESTADÍSTICAS BÁSICAS DE REGIONES ---")
display(df_regiones.describe(include='all'))
```

```python
# --- PUNTO 4: Creación de la variable IMC ---
df_atletas['IMC'] = df_atletas['Weight'] / ((df_atletas['Height'] / 100) ** 2)

print("--- ESTADÍSTICAS DEL IMC ---")
display(df_atletas[['Weight', 'Height', 'IMC']].describe())

imc_deporte = df_atletas.groupby('Sport')['IMC'].mean().dropna().sort_values()
print("\nTop 5 deportes con menor IMC promedio:\n", imc_deporte.head(5))
print("\nTop 5 deportes con mayor IMC promedio:\n", imc_deporte.tail(5))
```

```python
# --- PUNTO 5: Fusión (merge) de atletas con regiones usando NOC ---
filas_antes = len(df_atletas)
df_olimpiadas = pd.merge(df_atletas, df_regiones, on='NOC', how='left')
filas_despues = len(df_olimpiadas)

print(f"Filas antes del merge: {filas_antes}")
print(f"Filas después del merge: {filas_despues}")
print(f"Filas sin región asignada: {df_olimpiadas['region'].isna().sum()}")
print("\nCódigos NOC sin región en la tabla auxiliar:")
print(df_olimpiadas[df_olimpiadas['region'].isna()][['NOC', 'Team']].drop_duplicates())
```

```python
# --- PUNTO 6: Variable binaria medalla y comparación con atletas únicos ---
df_olimpiadas['medalla'] = df_olimpiadas['Medal'].notna().astype(int)

print("Conteo de la variable 'medalla' (0 = Sin medalla, 1 = Con medalla):")
print(df_olimpiadas['medalla'].value_counts())

print(f"\nTotal de filas (participaciones): {len(df_olimpiadas)}")
print(f"Total de atletas únicos (ID): {df_olimpiadas['ID'].nunique()}")
```

```python
# PUNTO 7: Representatividad histórica por época, temporada y sexo
print("--- PARTICIPACIÓN POR TEMPORADA (Season) ---")
display(df_olimpiadas['Season'].value_counts(dropna=False))
display(df_olimpiadas['Season'].value_counts(normalize=True) * 100)

print("\n--- PARTICIPACIÓN POR SEXO (Sex) ---")
display(df_olimpiadas['Sex'].value_counts(dropna=False))
display(df_olimpiadas['Sex'].value_counts(normalize=True) * 100)

print("\n--- PARTICIPACIÓN POR DÉCADA (Época) ---")
df_olimpiadas['Decada'] = (df_olimpiadas['Year'] // 10) * 10
resumen_epoca = df_olimpiadas.groupby(['Decada', 'Sex']).size().unstack(fill_value=0)
resumen_epoca['Total'] = resumen_epoca.sum(axis=1)
resumen_epoca['%_Mujeres'] = (resumen_epoca['F'] / resumen_epoca['Total']) * 100
display(resumen_epoca)
```

```python
# PUNTO 8: Análisis de valores faltantes y comparación antes vs. después de 1960
print("--- TOTAL Y PORCENTAJE DE DATOS FALTANTES POR VARIABLE ---")
faltantes = pd.DataFrame({
    'Faltantes': df_olimpiadas.isna().sum(),
    'Porcentaje (%)': (df_olimpiadas.isna().mean() * 100).round(2)
})
display(faltantes[faltantes['Faltantes'] > 0])

# Comparación antes de 1960 vs. 1960 en adelante
df_olimpiadas['Epoca_1960'] = df_olimpiadas['Year'].apply(
    lambda x: 'Antes de 1960' if x < 1960 else '1960 o posterior'
)

vars_evaluar = ['Age', 'Height', 'Weight', 'IMC', 'Medal', 'region']
faltantes_epoca = df_olimpiadas.groupby('Epoca_1960')[vars_evaluar].apply(
    lambda g: (g.isna().mean() * 100).round(2)
)
print("\n--- PORCENTAJE DE DATOS FALTANTES SEGÚN ÉPOCA (%) ---")
display(faltantes_epoca)
```

```python
# PUNTO 9: Respuestas directas (a - g)

# (a) Edad promedio de los atletas
edad_prom = df_olimpiadas['Age'].mean()

# (b) Registros con medalla vs. sin medalla
conteo_medallas = df_olimpiadas['medalla'].value_counts()

# (c) Estatura promedio de los atletas que ganaron medalla de oro
estatura_oro = df_olimpiadas[df_olimpiadas['Medal'] == 'Gold']['Height'].mean()

# (d) Ediciones de los Juegos (Games) y países (NOC) distintos
ediciones_distintas = df_olimpiadas['Games'].nunique()
noc_distintos = df_olimpiadas['NOC'].nunique()

# (e) Edad más joven y más vieja
edad_min = df_olimpiadas['Age'].min()
edad_max = df_olimpiadas['Age'].max()

# (f) Registros por temporada (Verano e Invierno)
reg_temporada = df_olimpiadas['Season'].value_counts()

# (g) Atletas únicos en 1 edición vs. más de 1 edición
ediciones_por_atleta = df_olimpiadas.groupby('ID')['Games'].nunique()
una_edicion = (ediciones_por_atleta == 1).sum()
mas_de_una = (ediciones_por_atleta > 1).sum()

print(f"(a) Edad promedio de los atletas: {edad_prom:.2f} años")
print(f"(b) Registros sin medalla (0): {conteo_medallas[0]} | Con medalla (1): {conteo_medallas[1]}")
print(f"(c) Estatura promedio de ganadores de Oro: {estatura_oro:.2f} cm")
print(f"(d) Ediciones distintas (Games): {ediciones_distintas} | Países distintos (NOC): {noc_distintos}")
print(f"(e) Edad más joven: {edad_min} años | Edad más vieja: {edad_max} años")
print(f"(f) Registros por temporada:\n{reg_temporada.to_string()}")
print(f"(g) Atletas únicos en 1 sola edición: {una_edicion} | En más de 1 edición: {mas_de_una}")
```

```python
# PUNTO 10: Proporción de medallas según Sexo, Temporada y Década

# 1. Proporción por Sexo
medallas_sexo = df_olimpiadas.groupby('Sex').agg(
    Total_Participaciones=('medalla', 'count'),
    Total_Medallas=('medalla', 'sum'),
    Proporcion_Medalla=('medalla', lambda x: round(x.mean() * 100, 2))
)
print("--- PROPORCIÓN DE MEDALLAS POR SEXO ---")
display(medallas_sexo)

# 2. Proporción por Temporada
medallas_temporada = df_olimpiadas.groupby('Season').agg(
    Total_Participaciones=('medalla', 'count'),
    Total_Medallas=('medalla', 'sum'),
    Proporcion_Medalla=('medalla', lambda x: round(x.mean() * 100, 2))
)
print("\n--- PROPORCIÓN DE MEDALLAS POR TEMPORADA ---")
display(medallas_temporada)

# 3. Proporción por Década
medallas_decada = df_olimpiadas.groupby('Decada').agg(
    Total_Participaciones=('medalla', 'count'),
    Total_Medallas=('medalla', 'sum'),
    Proporcion_Medalla=('medalla', lambda x: round(x.mean() * 100, 2))
)
print("\n--- PROPORCIÓN DE MEDALLAS POR DÉCADA ---")
display(medallas_decada)
```

```python
# PUNTO 11: Relación entre cantidad de pruebas por atleta y proporción de medalla

# Calcular cuántas pruebas (registros) disputó cada atleta único
df_olimpiadas['n_pruebas'] = df_olimpiadas.groupby('ID')['Event'].transform('count')

# Clasificar en los 3 grupos de participación
def clasificar_participacion(n):
    if n == 1:
        return '1 prueba'
    elif 2 <= n <= 3:
        return '2-3 pruebas'
    else:
        return '4 o más pruebas'

df_olimpiadas['grupo_participacion'] = df_olimpiadas['n_pruebas'].apply(clasificar_participacion)

# Ordenar categorías lógicamente y calcular proporción
orden_grupos = ['1 prueba', '2-3 pruebas', '4 o más pruebas']
resumen_pruebas = df_olimpiadas.groupby('grupo_participacion').agg(
    Registros=('medalla', 'count'),
    Atletas_Unicos=('ID', 'nunique'),
    Medallas=('medalla', 'sum'),
    Proporcion_Medalla=('medalla', lambda x: round(x.mean() * 100, 2))
).reindex(orden_grupos)

print("--- PROPORCIÓN DE MEDALLAS SEGÚN VOLUMEN DE PRUEBAS ---")
display(resumen_pruebas)
```

```python
# PUNTO 12: Tasa de medalla por grupo etario y comparación por temporada

# Crear rangos de edad
bins_edad = [0, 19, 24, 29, 34, 120]
etiquetas_edad = ['Menor de 20', '20-24 años', '25-29 años', '30-34 años', '35 o más']
df_olimpiadas['grupo_etario'] = pd.cut(df_olimpiadas['Age'], bins=bins_edad, labels=etiquetas_edad)

# 1. Proporción general por grupo etario
medallas_edad = df_olimpiadas.groupby('grupo_etario', observed=False).agg(
    Participaciones=('medalla', 'count'),
    Medallas=('medalla', 'sum'),
    Proporcion_Medalla=('medalla', lambda x: round(x.mean() * 100, 2))
)
print("--- PROPORCIÓN DE MEDALLAS POR GRUPO ETARIO ---")
display(medallas_edad)

# 2. Comparación entre Verano e Invierno
tabla_edad_temp = df_olimpiadas.pivot_table(
    index='grupo_etario',
    columns='Season',
    values='medalla',
    aggfunc=lambda x: round(x.mean() * 100, 2),
    observed=False
)
print("\n--- TASA DE MEDALLA (%) POR EDAD Y TEMPORADA ---")
display(tabla_edad_temp)
```
